# Uniform Cost Search for a Self-Driving Trailer

## 1. Introduction

This practical exercise implements Uniform Cost Search to determine the optimal route for a self-driving trailer travelling from Nairobi to Kisumu through the Kenyan highway network.
## 2. Problem Definition

- Initial State: Nairobi
- Goal State: Kisumu
- Search Algorithm: Uniform Cost Search
Path Cost: Distance travelled in kilometres
Objective: Find the route with the minimum total distance.

```python
import heapq
# Define the directed graph
graph = {
    "Nairobi": [("Naivasha", 92.8)],
    "Naivasha": [("Nakuru", 71.1)],
    "Nakuru": [
        ("Kericho", 107),
        ("Kisumu", 191)
    ],
    "Kericho": [("Kisumu", 83.9)],
    "Kisumu": []
}


# Uniform Cost Search algorithm
def uniform_cost_search(graph, start, goal):

    priority_queue = [(0, start, [start])]
    visited = {}

    while priority_queue:

        cost, city, path = heapq.heappop(priority_queue)

        if city in visited and visited[city] <= cost:
            continue

        visited[city] = cost

        print(f"Visiting: {city}, Cost: {cost} km")

        if city == goal:
            return path, cost

        for neighbour, distance in graph[city]:
            new_cost = cost + distance

        if neighbour not in visited or new_cost < visited[neighbour]:
                new_path = path + [neighbour]
                heapq.heappush(priority_queue, (new_cost, neighbour, new_path))

    return None, float("inf")


# Run the search
start = "Nairobi"
goal = "Kisumu"

optimal_path, total_cost = uniform_cost_search(graph, start, goal)


# Display results
print("\nOptimal Path:", " -> ".join(optimal_path))
print("Total Path Cost:", total_cost, "km")
```

## 3. Results and Interpretation

The Uniform Cost Search algorithm successfully found a path from Nairobi to Kisumu.

The algorithm explored the cities based on their cumulative travel costs.

### Optimal Path

Nairobi → Naivasha → Nakuru → Kericho → Kisumu

### Total Path Cost

354.8 km

### Explanation

The algorithm selected the route through Kericho because its total distance of 354.8 km is lower than the alternative route through Nakuru directly to Kisumu, which costs 354.9 km.

Therefore, the optimal route is Nairobi → Naivasha → Nakuru → Kericho → Kisumu.
